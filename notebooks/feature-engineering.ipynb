Feature Engineering for Fraud Detection

Load Data

```python
import pandas as pd
import numpy as np

f_df = pd.read_csv("../data/raw/Fraud_Data.csv")
```

Datetime Features

```python
f_df["signup_time"] = pd.to_datetime(f_df["signup_time"])
f_df["purchase_time"] = pd.to_datetime(f_df["purchase_time"])
```

Time Since Signup

```python
f_df["time_since_signup"] = (
    f_df["purchase_time"] - f_df["signup_time"]
).dt.total_seconds()
```

```python
Hour of Day
```

```python
f_df["hour_of_day"] = f_df["purchase_time"].dt.hour
```

Day of Week

```python
f_df["day_of_week"] = f_df["purchase_time"].dt.dayofweek
```

Transaction Velocity (User Level)

```python
user_freq = f_df.groupby("user_id").size().reset_index(name="txn_count")
f_df = f_df.merge(user_freq, on="user_id", how="left")
```

Device Velocity

```python
device_freq = f_df.groupby("device_id").size().reset_index(name="device_txn_count")
f_df = f_df.merge(device_freq, on="device_id", how="left")
```

Save Features

```python
f_df.to_csv("../data/raw/processed/fraud_features.csv", index=False)
```
